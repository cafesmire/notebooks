```python
df_april2019 = pd.read_csv('./data/SpotifyAudioFeaturesApril2019.csv')
df_nov2018 = pd.read_csv('./data/SpotifyAudioFeaturesNov2018.csv')
```

```python
df_april2019['popularity_2019'] = df_april2019['popularity']
df_april2019.drop(columns='popularity', inplace=True)

df_nov2018['popularity_2018'] = df_nov2018['popularity']
df_nov2018.drop(columns='popularity', inplace=True)
```

```python
df_april2019['track_id'].dtype
```

```python
df = df_nov2018.merge(df_april2019, on='track_id', how='outer', suffixes=['_2018', '_2019'])
df.head()
```

```python
df.info()
```

```python
df['popularity_difference'] = np.abs(df['popularity_2018'] - df['popularity_2019'])
df['popularity_difference']= df['popularity_difference'].fillna(0)
df['popularity_difference']
```
